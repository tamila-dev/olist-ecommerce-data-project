```python
import pandas as pd
from pathlib import Path

# Load reviews dataset
reviews = pd.read_csv(
    Path("data/raw/olist_order_reviews_dataset.csv")
)

# Find rows with duplicated review_id
duplicate_reviews = reviews[
    reviews.duplicated(subset=["review_id"], keep=False)
].sort_values("review_id")

# Display results
print("Total rows with duplicated review_id:", len(duplicate_reviews))
print("Number of duplicated review_id values:", duplicate_reviews["review_id"].nunique())

print("\nFirst 10 duplicated rows:")
print(duplicate_reviews.head(10).to_string(index=False))
```

```python
# Count orders associated with each review_id
orders_per_review = reviews.groupby("review_id")["order_id"].nunique()

print("Number of orders per review_id:")
print(orders_per_review.value_counts().sort_index())
```

```python
print(
    "Duplicate review_id + order_id combinations:",
    reviews.duplicated(
        subset=["review_id", "order_id"]
    ).sum()
)
```
